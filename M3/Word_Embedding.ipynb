# **단어 임베딩 (Word Embedding)**

---

## **1. 단어 임베딩의 정의**

- 기존의 텍스트 코딩 방식 (원핫 인코딩, BOW, 문서-단어 행렬 등)은  
  각 **단어마다 고유번호(Index)** 를 배정하여 사용함.
- 하지만 이 번호에는 **의미가 전혀 없으며**,  
  단순히 **인덱스 성격**만 가지고 있음.
- **단어 임베딩(Word Embedding)** 은 단어를 단순 인덱스가 아닌  
  **의미 있는 수치 벡터로 표현**하는 방법임.

---

## **2. 단어 벡터의 개념**

- **단어 벡터 (Word Vector)**  
  각 단어를 **50~300차원 정도의 벡터**로 표현함.

  예시  


학교 = [0.23, 0.58, 0.97, ..., 0.87, 0.95]

바다 = [0.45, 0.37, 0.81, ..., 0.22, 0.64]


- **단어 벡터의 특징**
- 단어 간의 **거리(distance)** 계산이 가능함.
- 거리 기반으로 **유의어/반의어** 탐색 가능.
- 성별, 단수/복수, 품사 구분 등의 관계를 학습 가능.
- 단, 각 차원의 값 자체에는 명확한 의미가 없음.

---

## **3. 단어 벡터의 학습**

- 단어 벡터는 **대형 말뭉치(Corpus)** 로부터 **학습**됨.
- 말뭉치의 문장들을 반복적으로 입력하여 학습시킴.
- 예를 들어 음식 관련 문장들을 학습시키면  
  음식 단어들 간의 유사한 벡터를 얻게 됨.

- **학습 예시 문장**
나는 어제 바나나를 맛있게 먹었다

이 망고는 먹기가 힘들다

이 사과는 씹는 맛이 아주 좋다

바나나가 사과보다 맛있다

잘 씹어야 맛있게 먹을 수 있다


- 위와 같은 문장들을 학습시키면  
‘바나나’, ‘사과’, ‘망고’, ‘포도’ 등의 단어가  
**의미적으로 가까운 벡터 공간상에 위치**하게 됨.

---

## **4. 사전 학습된 단어 벡터 (Pre-trained Embedding)**

- 이미 만들어진 임베딩을 가져다 사용할 수도 있음.

### **대표적인 예시: GloVe**

- **GloVe (Global Vectors for Word Representation)**  
- 2014년 스탠퍼드 대학에서 개발  
- 위키피디아 데이터로 학습  
- 약 **40만 개 단어를 100차원 벡터로 임베딩**

---

## **5. 단어 벡터의 연산적 특성**

- 단어 벡터 간의 관계는 **벡터 연산으로 표현 가능**함.

예시  
왕 : 여왕 = 아들 : ?

→ ? = 딸


- 이러한 관계는 다음과 같이 표현할 수 있음.
- (여왕 - 왕) + 아들 ≈ 딸
- 즉, 단어 간 의미적 차이를 **벡터의 방향과 거리**로 나타냄.

- 이런 관계를 시각화하면 다음과 같음.

| 관계 | 예시 |
|-------|------|
| 성별 관계 | 남자 → 여자, 왕 → 여왕 |
| 가족 관계 | 아들 → 딸 |
| 직업 관계 | 의사 → 간호사 |

---

## **6. 단어 임베딩의 핵심 요약**

| 구분 | 설명 |
|------|------|
| **정의** | 단어를 의미 있는 실수 벡터로 표현 |
| **목적** | 단어 간 유사도와 관계를 수치적으로 파악 |
| **학습 방법** | 대규모 말뭉치 기반 지도/비지도 학습 |
| **대표 모델** | Word2Vec, GloVe, FastText |
| **활용 분야** | 감정 분석, 문장 분류, 챗봇, 기계 번역 등 |

---

## **7. Word2Vec의 개념 요약**

- 구글에서 제안한 **단어 임베딩 학습 알고리즘**
- 단어의 주변 단어(Context)를 예측하거나,  
반대로 중심 단어를 예측하는 방식으로 학습함.

### **모델 구조**
| 구분 | 설명 |
|------|------|
| **CBOW (Continuous Bag of Words)** | 주변 단어(Context)로 중심 단어를 예측 |
| **Skip-gram** | 중심 단어로 주변 단어를 예측 |

### **예시**

문장: "나는 학교에 간다"

- CBOW: ["나는", "에", "간다"] → "학교"
- Skip-gram: "학교" → ["나는", "에", "간다"]


---

## **8. Word2Vec의 수학적 구조**

- **입력층(Input Layer):** 원-핫 인코딩된 단어
- **가중치 행렬(Weight Matrix):** 단어 임베딩 학습 대상
- **출력층(Output Layer):** Softmax를 통해 예측 확률 계산

학습 후 가중치 행렬이 바로 **단어 임베딩 벡터**로 사용됨.

---

## **9. 단어 임베딩의 확장**

- **FastText**
  - 단어 내부의 n-gram 단위까지 학습하여  
    **형태소 기반 단어 표현** 가능
  - 예: “playing” → “play”, “lay”, “ying” 포함

- **ELMo / BERT**
  - 문맥(Context)에 따라 단어의 의미가 달라지는  
    **문맥적 임베딩(Contextual Embedding)** 방식

---

## **10. 정리 요약**

| 구분 | 내용 |
|------|------|

| **임베딩 이후** | 단어가 의미 벡터로 표현 (의미 유사성 표현 가능) |

# **문장 유사도**
- ### 편집 거리를 이용한 방법
- ### 단어 벡터을 이용한 방법

### **단어의 유사도**
- #### 두 개의 문자열이 얼마나 다른지를 나타내는 편집 거리를 이용 
- #### 편집 거리 
    - 한 단어에서 글자를 추가, 제거, 변경함으로써 다른 단어로 바꿀 때 필요한 최소한의 편집 행동의 횟수로 결정
    - NLTK 라이브러리를 활용하여 두 문장의 편집거리 계산이 가능하
        - nltk.edit_distance(i, j)

```python
import nltk

sentence_list = ["우리 모두 함께 놀자", "모두 같이 놀자", "놀자", "모두 다 같이"]

for i in sentence_list:
    print("'", i, "'")
    
    for j in sentence_list:
        print("\t", j, " : ", end='')
        print(nltk.edit_distance(i, j))
        
    print()
```

### **코사인 유사도- 직접 구현**

```python
# 직접 구현시 코사인 유사도
import math
def cosine_similarity(v1,v2):
    
    sumxx, sumxy, sumyy = 0, 0, 0
    for i in range(len(v1)):
        x = v1[i]; y = v2[i]
        sumxx += x*x
        sumyy += y*y
        sumxy += x*y
    return sumxy/math.sqrt(sumxx*sumyy)
```

### **자카드 유사도- 직접 구현**
- #### str1.split()
- #### a.intersection(b)

```python
# 자카드 유사도 계산

sentence_list = ['우리 모두 함께 놀자','모두 함께 놀자']

def get_jaccard_sim(str1, str2): 
    a = set(str1.split()) 
    b = set(str2.split())
    c = a.intersection(b)
    return float(len(c)) / (len(a) + len(b) - len(c))
```

```python
get_jaccard_sim(sentence_list[0], sentence_list[1])
```
