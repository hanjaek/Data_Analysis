# **텍스트 분석 (Text Mining / NLP)**

---

## **1. 텍스트 분석의 개요**

- **텍스트 분석의 목적**
  - 텍스트의 **의미를 알아내는 것**
  - 글의 목적, 글쓴이의 성향(찬성/반대), 기분(기쁨/슬픔/우울함), 제품 피드백 등

- **특징**
  - 텍스트는 대표적인 **비정형 데이터**
  - 의미를 추출하려면 비정형 데이터에서 **정형화된 정보**를 얻어야 함
  - 텍스트의 구문을 **분석하여 의미를 파악**하고 이를 **정량적으로 측정**

---

## **2. 텍스트 분석의 활용**

- **SNS 분석 (트위터, 블로그, 페이스북 등)**
  - 소비자의 반응, 감성, 트렌드를 파악
  - 개인별 마케팅 및 상품 피드백 분석

- **문서 분석 (이메일, 댓글, 뉴스 기사, 콜센터 상담기록 등)**
  - 주요 내용 및 특징을 추출
  - 유사한 글이나 저자 탐색에 활용

- **연구 및 인용 네트워크 분석**
  - 문서 간 관계 및 전문가 간의 인적 네트워크 파악

- **AI 응용**
  - 인공지능 스피커, **챗봇(Chatbot)** 등에서 기본적으로 텍스트 분석 기술 활용

---

## **3. 텍스트 표현 방법**

- 컴퓨터는 사람처럼 단어나 문장의 의미를 직접 이해할 수 없음  
  → **비정형 데이터 → 정형 데이터(수치형)** 로 변환 필요

### **텍스트 데이터 처리 과정**
1. **비정형 데이터:** 원문 텍스트
2. **정형화 단계:** 토큰화, 정제, 불용어 제거 등
3. **수치 데이터 변환:** One-hot, TF-IDF, Word2Vec 등 벡터화 방식 적용

---

## **4. 코퍼스(Corpus, 말뭉치)**

- **말뭉치(Corpus):** 분석 대상이 되는 전체 문서의 집합
- **문서(Document):** 코퍼스 내 한 단위의 텍스트
  - 예시) 하나의 블로그 = 문서  
    블로그 1000개 = 말뭉치
- **파싱(Parsing):** 코퍼스에서 **의미 있는 단어를 추출**하는 작업

---

## **5. 토큰화(Tokenization)**

- **의미:** 텍스트를 컴퓨터가 이해할 수 있는 최소 단위인 **토큰(token)** 으로 분리
- **단계:** 텍스트 분석의 첫 단계

### **토큰 단위 종류**
| 구분 | 설명 |
|------|------|
| 단어(word) | 일반적인 텍스트 분석에서 가장 많이 사용 |
| 문자(character) | 한 글자씩 구분 |
| n-gram | 연속된 n개의 단어를 묶음으로 사용 |

### **주의점**
- 같은 의미의 단어라도 형태가 다르면 다른 토큰으로 인식됨  
  → **어간추출(stemming)** 또는 **표제어추출(lemmatization)** 으로 해결

---

## **6. 원-핫 인코딩 (One-Hot Encoding)**

- 토큰에 고유 번호를 부여하고, 해당 위치만 1로 표시하는 벡터 방식

예시  
문장: "어제 러시아에 갔다가 러시아 월드컵을 관람했다"

토큰 사전: { "어제":0, "러시아":1, "갔다":2, "월드컵":3, "관람":4 }

- 어제 → [1, 0, 0, 0, 0]
- 러시아 → [0, 1, 0, 0, 0]
- 갔다 → [0, 0, 1, 0, 0]
- 월드컵 → [0, 0, 0, 1, 0]
- 관람 → [0, 0, 0, 0, 1]


- 장점: 단순하고 이해하기 쉬움  
- 단점: 벡터 차원이 커지고, 단어 간 의미 관계를 표현하기 어려움

---

## **7. TF-IDF (단어 중요도 기반 표현)**

> *단어의 빈도와 희귀도를 함께 고려한 가중치 방식*

- **TF (Term Frequency):** 단어가 문서에 얼마나 자주 등장했는가  
- **IDF (Inverse Document Frequency):** 전체 문서에서 얼마나 희귀한가  
- **TF-IDF = TF × IDF**

| 단어 | TF | IDF | TF-IDF |
|------|----|-----|--------|
| 영화 | 0.3 | 1.2 | 0.36 |
| 좋다 | 0.5 | 0.8 | 0.40 |
| 그리고 | 0.7 | 0.2 | 0.14 |

---

## **8. 형태소 분석 & 불용어 제거**

- **형태소 분석:** 문장을 품사 단위로 분리 (명사, 동사, 조사 등)  
- **불용어 제거(Stopwords):** 의미 없는 단어 제거 (예: "그리고", "그는", "이것은" 등)

### **문서-단어 행렬 만들기**
- 전체 문서 {d1,d2,…,dn}를 구성하는 고정된 단어장(vocabulary) {t1,t2,…,tm}를 만든다.
- 각 di 개별 문서에 단어장에 해당하는 단어들이 포함되어 있는지를 표시한다.
- x(i,j) = 문서 di 내의 단어 tj의 출현 빈도
- 또는
- x(i,j) = {0,1}
    - 단어 tj가 문서 di 안에 없으면 : 0
    - 단어 tj가 문서 di 안에 있으면 : 1

### **CountVectorizer()**
- 1. 문서를 토큰(단어) 리스트로 변환한다.
- 2. 각 문서에서 토큰(단어)의 출현 빈도를 센다. : vocabulary_
- 3. 각 문서를 BOW 인코딩 벡터로 변환한다. --> tf(term frequency) matrix

#### **(인수 설명)**
- stop_words : 문자열 {‘english’}, 리스트 또는 None (디폴트)
    - stop words 목록.‘english’이면 영어용 스탑 워드 사용
- ngram_range : (min_n, max_n) 튜플
    - n-그램 범위
- max_df : 정수 또는 [0.0, 1.0] 사이의 실수. 디폴트 1
    - 단어장에 포함되기 위한 최대 빈도
- min_df : 정수 또는 [0.0, 1.0] 사이의 실수. 디폴트 1
    - 단어장에 포함되기 위한 최소 빈도

### **TF-IDF(Term Frequency – Inverse Document Frequency) 인코딩**
- 단어를 갯수 그대로 카운트하지 않고 모든 문서에 공통적으로 들어있는 단어의 경우 문서 구별 능력이 떨어진다고 보아 가중치를 축소하는 방법
- 구제적으로는 문서 d(document)와 단어 t 에 대해 다음과 같이 계산
- tf-idf(d,t)=tf(d,t)⋅idf(t)
    - tf(d,t) : term frequency. 
        - 특정한 단어의 빈도수
    - idf(t)  : inverse document frequency. 
        - 특정한 단어가 들어 있는 문서의 수에 반비례하는 수

- idf(d,t)=log(n/(1+df(t)))
- n  : 전체 문서의 수
- df(t) : 단어 t 를 가진 문서의 수

````python
tv = TfidfVectorizer(max_features=2000)
dtm = tv.fit_transform(news_text.tolist())
df = pd.DataFrame(dtm.toarray(), columns = tv.get_feature_names_out())
df.head(5)
````
